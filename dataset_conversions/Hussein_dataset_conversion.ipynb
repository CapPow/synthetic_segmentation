Organize Hussein external validation set for use

```python
ext_output_dir = "/run/media/john/Storage/herb_semantic_segmentation/validation sources/Hussein_dataset"
ext_img_path = "./Hussein_images"
ext_mask_path = "./Hussein_masks"
ext_base_names = [os.path.basename(x).split(".")[0] for x in glob(f"{ext_output_dir}/images/*.png")]
# filter previously converted ext_base_names 
existing_ext_base_names = [os.path.basename(x).split("_")[0] for x in glob(f"{ext_img_path}/*.jpg")]
ext_base_names = [x for x in ext_base_names if x not in existing_ext_base_names]

def ext_copy_to_dataset(base_name, dst_img_path, dst_mask_path):
    src_img = f"{ext_output_dir}/images/{base_name}.png"
    dst_img = f"{dst_img_path}/{base_name}_src.jpg"
    shutil.copy2(src_img, dst_img)

    src_mask = cv2.imread(f"{ext_output_dir}/ground_truth/{base_name}.png", cv2.IMREAD_GRAYSCALE)
    # convert the leaves to 6, and everything else to 0
    src_mask[src_mask == 90 ] = 6
    src_mask[src_mask != 6] = 0
    dst_mask = f"{dst_mask_path}/{base_name}_msk.png"
    # explicitly use no compression for the mask
    cv2.imwrite(dst_mask, src_mask, [cv2.IMWRITE_PNG_COMPRESSION, 0])

for base_name in ext_base_names:
    ext_copy_to_dataset(base_name, ext_img_path, ext_mask_path)
```
